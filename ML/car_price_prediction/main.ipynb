```python
X = df.drop(columns= ['price'])
y = df['price']
X
```

```python
X_one_encoded = pd.get_dummies(X , columns = ['model', 'transmission', 'fuelType'], drop_first = True)
```

```python
X_one_encoded = X_one_encoded.astype(int)
```

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

columns = ['model', 'transmission', 'fuelType']

Xlable = X.copy()

for i in columns:
    Xlable[i] = encoder.fit_transform(Xlable[i])

Xlable
```

```python
X.model.value_counts()
```

```python
from sklearn.preprocessing import StandardScaler

numerical_columns = ['year', 'mileage', 'tax', 'mpg', 'engineSize']

scaler = StandardScaler()
X_one_encoded[numerical_columns] = scaler.fit_transform(X_one_encoded[numerical_columns])
```

```python
X_one_encoded
```

```python
X.columns
```

```python
Xlable[['model', 'year', 'transmission', 'mileage', 'fuelType', 'tax', 'mpg',
       'engineSize']] = scaler.fit_transform(Xlable[['model', 'year', 'transmission', 'mileage', 'fuelType', 'tax', 'mpg',
       'engineSize']])
Xlable
```

### two seprate models for Xlable and X_one_encoded 

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_one_encoded, y, test_size=0.20, random_state=42)
```

```python
model = LinearRegression()
model.fit(X_train,y_train)
```

```python
y_predict = model.predict(X_test)
```

```python
y_predict
```

```python
r2 = r2_score(y_test, y_predict)
r2
```

```python
n = X_test.shape[0]
p = X_test.shape[1]
adjusted_r2 = 1-((1-r2)*(n-1))/(n-p-1)
adjusted_r2
```

```python
X_train, X_test, y_train, y_test = train_test_split(Xlable, y, test_size=0.20, random_state=42)

model = LinearRegression()
model.fit(X_train,y_train)

y_predict = model.predict(X_test)
```

```python
r2 = r2_score(y_test, y_predict)
r2

```

```python
n = X_test.shape[0]
p = X_test.shape[1]
adjusted_r2 = 1-((1-r2)*(n-1))/(n-p-1)
adjusted_r2
```
